```python
is_var = lambda x: isinstance(x, str) and x[0].islower()

def occurs(v, term, sub):
    if v == term: return True
    if is_var(term) and term in sub: return occurs(v, sub[term], sub)
    if isinstance(term, tuple): return any(occurs(v, arg, sub) for arg in term[1:])
    return False

def unify(x, y, sub=None):
    if sub is None: sub = {}
    if x == y: return sub
    if is_var(x): return unify_var(x, y, sub)
    if is_var(y): return unify_var(y, x, sub)
    if isinstance(x, tuple) and isinstance(y, tuple) and x[0] == y[0] and len(x) == len(y):
        for a, b in zip(x[1:], y[1:]):
            if (sub := unify(a, b, sub)) is None: return None
        return sub
    return None

def unify_var(v, x, sub):
    if v in sub: return unify(sub[v], x, sub)
    if is_var(x) and x in sub: return unify(v, sub[x], sub)
    if occurs(v, x, sub): return None
    return {**sub, v: x}

# --- Quick Verification ---
e1 = ('Knows', 'John', 'x')
e2 = ('Knows', 'y', 'jane')
print("MGU:", unify(e1, e2))  # Output: {'y': 'John', 'x': ('Mother', 'y')}
```
