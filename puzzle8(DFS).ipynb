```python
import copy

# Define the target configuration (Goal State)
# 0 represents the empty/blank space
GOAL_STATE = (
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 0)
)

def find_blank(state):
    """Finds the (row, col) coordinates of the blank tile (0)."""
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                return r, c

def get_neighbors(state):
    """Generates all valid next states by shifting the blank tile."""
    neighbors = []
    r, c = find_blank(state)

    # Define possible movements: (row_change, col_change, action_name)
    moves = [
        (1, 0, 'Down'),
        (-1, 0, 'Up'),
        (0, -1, 'Left'),
        (0, 1, 'Right')
    ]

    for dr, dc, action in moves:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            # Convert tuple of tuples to list of lists to allow mutation
            new_state = [list(row) for row in state]
            # Swap the blank tile with the target tile
            new_state[r][c], new_state[nr][nc] = new_state[nr][nc], new_state[r][c]
            # Convert back to immutable tuple format for set tracking
            neighbors.append((tuple(tuple(row) for row in new_state), action))

    return neighbors

def solve_8_puzzle_dfs(start_state):
    """Solves the 8-puzzle using an iterative Depth-First Search strategy."""
    # Stack stores tuples of: (current_state, path_taken_to_get_here)
    stack = [(start_state, [])]
    visited = set()

    while stack:
        current_state, path = stack.pop()

        # Check if we reached our target
        if current_state == GOAL_STATE:
            return path

        if current_state not in visited:
            visited.add(current_state)

            # Explore neighbors.
            # Reversing neighbors forces a specific traversal order (e.g., Up, Down, Left, Right)
            for neighbor_state, action in reversed(get_neighbors(current_state)):
                if neighbor_state not in visited:
                    new_path = path + [action]
                    stack.append((neighbor_state, new_path))

    return None # Return None if no solution is found

def print_board(state):
    """Helper function to print the 3x3 matrix beautifully."""
    for row in state:
        print(" ".join(str(tile) if tile != 0 else "_" for tile in row))
    print()

# =====================================================================
# Example Execution
# =====================================================================
if __name__ == "__main__":
    # Define a simple starting board layout (Solvable in 3 moves)
    initial_state = (
        (1, 2, 3),
        (4,5,0),
        (7,8,6)
    )

    print("--- Initial State ---")
    print_board(initial_state)

    print("Searching for solution via DFS...")
    solution_path = solve_8_puzzle_dfs(initial_state)

    if solution_path:
        print(f"Solution found in {len(solution_path)} moves!")
        print(f"Sequence of actions: {' -> '.join(solution_path)}")
    else:
        print("No solution found or state configuration is unsolvable.")
```
