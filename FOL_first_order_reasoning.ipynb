```python

facts = {
    ("Owns", "A", "T1"),
    ("Missile", "T1"),
    ("American", "Robert"),
    ("Enemy", "A", "America")
}

rules = [
    (
        [("Missile", "?x"), ("Owns", "A", "?x")],
        ("Sells", "Robert", "?x", "A")
    ),
    (
        [("Missile", "?x")],
        ("Weapon", "?x")
    ),
    (
        [("Enemy", "?x", "America")],
        ("Hostile", "?x")
    ),
    (
        [("American", "?p"), ("Weapon", "?q"),
         ("Sells", "?p", "?q", "?r"), ("Hostile", "?r")],
        ("Criminal", "?p")
    )
]


def match(pattern, fact, substitution):
    if pattern[0] != fact[0] or len(pattern) != len(fact):
        return None

    new_sub = substitution.copy()

    for p, f in zip(pattern[1:], fact[1:]):
        if p.startswith("?"):
            if p in new_sub:
                if new_sub[p] != f:
                    return None
            else:
                new_sub[p] = f
        elif p != f:
            return None

    return new_sub


def forward_chain(facts, rules):
    changed = True

    while changed:
        changed = False

        for conditions, conclusion in rules:

            substitutions = [{}]

            for condition in conditions:
                new_substitutions = []

                for sub in substitutions:
                    for fact in facts:
                        result = match(condition, fact, sub)

                        if result is not None:
                            new_substitutions.append(result)

                substitutions = new_substitutions

            for sub in substitutions:
                new_fact = tuple(
                    sub.get(x, x) for x in conclusion
                )

                if new_fact not in facts:
                    facts.add(new_fact)
                    print("Derived:", new_fact)
                    changed = True

    return facts


facts = forward_chain(facts, rules)

query = ("Criminal", "Robert")

if query in facts:
    print("\nRobert is Criminal: TRUE")
else:
    print("\nRobert is Criminal: FALSE")
```
