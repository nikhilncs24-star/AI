```python
import itertools

# Define logical implication (A -> B is equivalent to not A or B)
def implies(a, b):
    return (not a) or b

# Helper to format boolean values as 'T' or 'F'
def fmt(val):
    return 'T' if val else 'F'

# Print table header
header = f"{'p':<3} | {'q':<3} | {'r':<3} | {'q->p':<5} | {'p->~q':<6} | {'q v r':<6} | {'KB':<3}"
print(header)
print("-" * len(header))

# Generate all 8 truth combinations (True first to match standard table order)
for p, q, r in itertools.product([True, False], repeat=3):
    q_imp_p = implies(q, p)
    p_imp_not_q = implies(p, not q)
    q_or_r = q or r
    kb = q_imp_p and p_imp_not_q and q_or_r

    print(f"{fmt(p):<3} | {fmt(q):<3} | {fmt(r):<3} | {fmt(q_imp_p):<5} | {fmt(p_imp_not_q):<6} | {fmt(q_or_r):<6} | {fmt(kb):<3}")
```
