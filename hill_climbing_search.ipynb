```python
import random

def get_h_score(board):
    """
    Calculates the heuristic cost (h-score): the number of mutually attacking pairs of queens.
    board: list of integers where index=column and value=row.
    """
    n = len(board)
    attacks = 0
    for i in range(n):
        for j in range(i + 1, n):
            # Same row conflict or diagonal conflict
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                attacks += 1
    return attacks

def generate_all_neighbors(board):
    """
    Generates every possible neighbor state by moving exactly one queen
    to a different row within its current column.
    Returns a list of tuples: (neighbor_board_state, its_h_score)
    """
    n = len(board)
    neighbors = []

    for col in range(n):
        original_row = board[col]
        for row in range(n):
            if row == original_row:
                continue  # Skip the current position

            # Create a copy of the board and move the queen
            neighbor_board = list(board)
            neighbor_board[col] = row

            # Calculate heuristic for this specific neighbor
            h_score = get_h_score(neighbor_board)
            neighbors.append((neighbor_board, h_score))

    return neighbors

def hill_climbing_4_queens(max_restarts=100):
    """
    Solves the 4-Queens problem using Steepest-Ascent Hill Climbing with Random Restarts.
    Uses generate_all_neighbors to scan the full local neighborhood.
    """
    n = 4

    for restart in range(max_restarts):
        # 1. Generate a random initial complete-state configuration
        board = [random.randint(0, n - 1) for _ in range(n)]
        current_h = get_h_score(board)

        while True:
            # If no conflicts exist, we have found our global optimum
            if current_h == 0:
                return board

            # 2. Generate all neighbors
            neighbors = generate_all_neighbors(board)

            # 3. Find the lowest h-score among all generated neighbors
            min_neighbor_h = min(neighbor[1] for neighbor in neighbors)

            # 4. If the best neighbor doesn't improve our score, we hit a local optimum/plateau
            if min_neighbor_h >= current_h:
                break  # Trigger a random restart

            # 5. Gather all neighbors that share this best minimum score
            best_neighbors = [neighbor[0] for neighbor in neighbors if neighbor[1] == min_neighbor_h]

            # 6. Step into one of the best neighbors randomly
            board = random.choice(best_neighbors)
            current_h = min_neighbor_h

    return None

def print_board(board):
    """Utility to print the board layout cleanly."""
    if not board:
        print("No solution found.")
        return
    n = len(board)
    for row in range(n):
        line = ""
        for col in range(n):
            line += " Q " if board[col] == row else " . "
        print(line)

# Execution block
if __name__ == "__main__":
    print("Finding a solution for 4-Queens using Hill Climbing with explicit neighborhood generation...\n")
    solution = hill_climbing_4_queens()

    if solution:
        print(f"Solution state vector (rows per column): {solution}\n")
        print_board(solution)
    else:
        print("Failed to find a solution within the restart limit.")
```
